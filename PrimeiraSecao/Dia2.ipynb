# Trabalhando com bibliotecas

Para instalar uma bilbioteca, podemos instalá-las com o comando pip, do mesmo jeito que fizemos com o comando "pip", da mesma forma que fizemos com o Jupyter
Podemos instalar bibliotecas através do jupyter colocando uma exclamação(!) antes do comando

Após instalada, podemos importar as bibliotecas para utilizar suas funcionalidades com o comando import seguido pelo nome da biblioteca.

```
import biblioteca
```

Caso você queira dar um "apelido" para a biblioteca para não precisar chamar seu nome, podemos usar a palavra reservada "as"

```
import biblioteca as bib
```

Nesse caso, estariamos importando a biblioteca "biblioteca", porém, para utilizarmos, usaríamos "bib" como nome

# Matplotlib
A primeira biblioteca que vamos aprender é a Matplotlib, biblioteca que permite a criação de gráficos

A principal parte do matplotlib que vamos utilizar se chama "pyplot". Vamos primeiro importar esse módulo para poder usar. Vamos usar o _alias_ de "plt" para não ter que escrever "matplotlib.pyplot" toda vez que formos utilizar.
```
import matplotlib.pyplot as plt 
```

Agora, com o módulo importado, podemos criar nosso primeiro gráfico. Para isso, o módulo pyplot tem a função "plot" que cria um gráfico simples.

```
lista = [1,2,3,1,2,7,3,6,5,8,4,13,8,3]
plt.plot(lista)
```
**O código acima cria o seguinte gráfico:**




A função usada do jeito que usamos acima, coloca o valor da lista como Y e a posição como x. Caso queiramos queiramos utilizar uma lista como x e outra como y, basta utilizar uma lista como x e uma como y. A primeira lista passada para a função representa os "x" e a segunda representa os "y"

Por exemplo, o seguinte código:

```
listaX = [1,2,3, 3.2, 3.4, 3.8, 4]
listaY = [1,2,7, 3,   9,   8,   13]
plt.plot(listaX, listaY)
```
