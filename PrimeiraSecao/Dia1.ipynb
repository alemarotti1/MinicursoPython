# Trabalhando com listas

Tipos de lista


### Lista(list):
- Muito útil quando precisamos de um agrupamento que podem ou não ser ordenados mas que pode ser repetido mas não precisa de nenhuma tag associado a ela.
    - Por exemplo, podemos usar listas quando queremos armazenar uma lista de alturas de uma sala, mas não queremos associar um identificador a cada item(como o nome)
    - Listas também podem ser usadas quando a ordem importa e a quantidade de itens varia(exemplo: itens de compra selecionados por usuários)

#### **Criando uma lista**:
Podemos criar uma lista de itens da seguinte forma:
```
lista2 = [item1, item2, item3]
```
Os itens da lista podem ter tipos variados, então podemos criar listas mistas, como a lista a seguir:

```
lista2 = [ "texto", 12, ["uma lista dentro de outra lista"] ]
```


#### **Utilizando listas**

1. Com listas criadas nós podemos acessar os valores da lista através da posição dos mesmos, da seguinte forma:
```
resultado = lista2[0] # As posições da lista começam em 0 e vão até (o tamanho da lista - 1)
print(resultado) # O código exibirá "texto" como resultado(se considerarmos a variável "lista2" criada acima)
```

2. Também podemos utilizar valores negativos como posições. Esses valores começam a contar a partir do fim da lista

```
resultado = lista2[-1]
print(resultado) # O código exibirá ["uma lista dentro da outra"]
```

3. O python também permite acessasr uma sequência de valores de uma lista ao mesmo tempo. Podemos fazer isso da seguinte forma:

```
sequencia = lista[inicio:quantidadeDeItens]
```

Por exemplo, o seguinte código retornaria as posições 0, 1 e 2
```
lista = [0,1,2,3,4]
resultado = lista[0:3]
```

Caso queiramos que a lista comece do início, ou vá até o final e não sabemos qual a posição queremos, podemos utilizar o dois pontos(:) sem um número do lado:

```
lista = [0,1,2,3,4]
resultado = lista[2:] # essa lista conterá [2,3,4]
```



4. Além disso, podemos obter quantos itens a lista possui com o comando len
```
tamanho = len(lista2)
print(tamanho) # Exibirá 3
```

5. Podemos adicionar itens na lista com a função append:

```
lista2.append(item5)
```

6. Podemos remover itens do final da lista usando o pop
```
lista3 = [1, 2, 3, 4]
lista3.pop() # lista3 => [1, 2, 3]
```
Também podemos passar a posição que desejamos apagar para o pop, para apagar um item específico
```
lista3.pop(0) # lista3 => [2, 3]
```

A função pop também retorna o item removido.


7. Podemos também juntar duas listas com a função "extend"

```
lista4 = [9,8,7]
lista3.extend(lista4) # lista3=> [2, 3, 9, 8, 7]  lista4=> [9, 8, 7]
```
Caso não queiramos alterar a lista atual, nós podemos utilizar o operador de soma, da seguinte forma

```
lista_conjunta = lista3+lista4
```

### Tuplas(tuple):
- Múito útil quando temos uma quantidade fixa de itens e sabemos que eles vão sempre estar presentes e a ordem vai ser sempre a mesma(ex, uma função de saldo do banco que em todas as requisições vão ser retornadas tanto o saldo atual como a variação diária da conta) ou uma posição x,y, z imutável. 

#### Criando uma tupla:
Podemos criar uma tupla de forma bem semelhante à forma que criamos uma lista, porém ao invés de colchetes - [] - usamos parêntesis - () 

```
tupla = (1, 2, 3)
```


#### Utilizando uma tupla

1. Acessando uma tupla:

Podemos acessar uma tupla da mesma forma que fazemos com uma lista:
```
resultado = tupla[0] #resultado => 1
```


2. Adicionando, alterando e exluindo:
Tuplas são imutáveis após criadas, então não podemos fazer nenhuma das operações que fazemos com as listas

3. Expansão de tuplas

Podemos expandir os itens de uma tupla, caso saibamos a quantidade de valores que ela possui da seguinte forma:

```
pos = (1, 2, 3)
(posX, posY, posZ) = pos
```

Caso não saibamos quantos itens a tupla, podemos acessar expandir a tupla da seguinte forma:
```
tuplaGrande = (1,2,3,4,5,6,7,8,9)
(valor1, valor2, expansao*) = tuplaGrande
```

Todos os valores da tupla que não couberem nos itens sem * irão para o "expansao". Vale a pena notar que o item com * pode estar em qualquer item da tupla que está sendo expandida

### Range(range):
- Coleção sequencial de números; imutável. Coleção muito útil quando estramos trabalhando com lações de repetição . Ela representa um conjunto sequencial de números em um determinado intervalo. 


#### Criando um ranges:
Podemos criar ranges da seguinte forma:
```
#sequencia = range(qtd_itens)

Além disso, se quisermos um salto diferente de 1, podemos fazer isso da seguinte forma:

Também podemos ter um range de trás pra frente, usando um salto negativo, da seguinte forma:

```

## Verificação de pertencimento

Muitas vezes gostariamos de verificar se um item pertence a uma lista ou não. Para isso, podemos utilizar os verificadores de pertencimento "in"(verifica se pertence) ou "not in"(que citamos anteriormente)

Para isso basta escrever o valor que queremos verificar, seguido do agrupamento que gostariamos de verificar

````python
print("10 in [10]:")
result = 10 in [10]
print(result)

print("\n10 in (10, 0):")
result = 10 in (10, 0)
print(result)

print("\n0 in range(10):")
result = 0 in range(10)
print(result)


# Em dicionários, verifica apenas a chave
print("\n10 in {\"chave\": 10}:")
result = 10 in {"chave": 10}
print(result)

print("\n\"chave\" in {\"chave\": 10}:")
result = "chave" in {"chave": 10}
print(result)

# Também funciona com variáveis
print("\ntestar_pertencimento in lista:")
testar_pertencimento = 10
lista = [0, 10]
result = testar_pertencimento in lista
print(result)

````

## Condicionais:
Em python, nós possuímos 3 tipos de condicionais mais comuns, que são o if, o elif e o else. Não é necessário parentesis para a execução do condicional

### if
Comando condicional base. Executa o bloco de código identado apenas caso a condição seja verdadeira, diferente de 0 ou diferente de None

```
if True:
    print("executou 1") # vai exibir "executou 1"

if False:
    print("executou 2") # não vai exibir 

if -1: # Valor diferente de zero - negativo também conta
    print("executou 3") # vai exibir "executou 3"

if 0:
    print("executou 4") # não vai exibir 

if None: 
    print("executou 5") # não vai exibir
```

### elif

Muitas vezes, apenas uma condição não é suficiente para avaliar todos os caminhos que podemos tomar. Para avaliar outras condições caso o primeiro "if" falhe, podemos utilizar o "elif". 

Nota: Vale notar que caso o código a ser executado dentro da condicional for o mesmo no if e no elif, não vale a pena usar um elif. Use apenas o if juntando as condicionais com um "or"

```
if False:
    pass # Comando "pass" significa "faça nada"

elif True: 
    print("entrei no elif") # vai exibir "entrei no elif"

```

Podemos utilizar quantos elif em sequência desejarmos:

```
if False:
    pass 

elif False:
    pass

elif True: 
    print("entrei no último elif") 

```

### else

Para executar um bloco de código caso todas as outras condições falhem, podemos usar o else como comando final:
```
if False:
    pass 

elif False: 
    pass

else:
    print("entrou no else")
```



Somente um "else" por bloco condicional pode ser utilizado

## Laços de repetição:

### laços for
Podemos utilizar o laço for para percorrer itens de um agrupamento(ou algo semelhante)
```
lista = ["pessoa1", "pessoa2", "pessoa3"]
for pessoa in lista:
    print(pessoa)
```

Caso queiramos executar um número específico de vezes, podemos utilizar o for em conjunto com o range
```
for i in range(10): # executará 10 vezes, de 0 a 9 
    print(i)
```

### Laços while
Quando queremos executar até uma condição ser atingida, ao invés de percorrer uma lista ou uma quantidade específica de vezes, podemos usar o while:

```
numero_inicial = 100; 
while numero_inicial % 77 != 0: # executa até encontrar um número divisível por 77
    numero_inicial += 1
print(numero_inicial)
```

### O laço de repetição disfarçado: list comprehension
Além dos loops mais comuns das outras linguagens de programação, o python possui um tipo de laço para a criação de listas a partir de outras listas. Ainda é um laço e ainda precisa percorrer a lista inteira, mas deixa o código muito mais legível

O list comprehension permite com que a gente possa pegar os itens de uma lista e fazer alguma alteração nele para criar uma outra lista 

```
lista_de_numeros=[1,2,3,4]
lista_dobrada=[x*2 for x in lista_de_numeros]
print(lista_dobrada) # lista vai ser [2, 4, 6, 8]
```

Além da alteração de itens da lista, também podemos colocar uma condicional para pegar apenas itens que cumprem determinada condição
```
lista_de_nomes=["Maria da Silva", "Jorge Faria", "Ana Melo Silva", "Marcio da Costa", "Pedro Lima Silva"]
nomes_silva=[x for x in lista_de_nomes if x[-5:]=="Silva"] # cria uma lista com os nomes terminados em "Silva"
print(nomes_silva) # lista vai ser ['Maria da Silva', 'Ana Melo Silva', 'Pedro Lima Silva']
```

# Funções

Para criar funções no python, utilizamos a palavra de definições "def" seguido pelo nome da função, os parametros separados por virgula dentro de parentesis e dois pontos(:). Novamente, todo o código dentro do bloco da função precisa ser identado

Caso a gente queira retornar alguma coisa, nós utilizamos a palavra "return" seguido pelo valor a retornar, como a maioria das linguagens de programação.


```
def funcao_exemplo():
    return "oi"
```

Para chamar a função, basta colocar o nome da função seguido de parentesis(com os parametros dentro separados por vírgula, caso eles existam)

```
valor_retorno = funcao_exemplo()
print(valor_retorno)
```

Caso a função tenha parametros, fica assim:
```
def funcao_com_parametros(parametro1, parametro2):
    print("os parametros", parametro1, parametro2)
```
