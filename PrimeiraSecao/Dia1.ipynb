## Variáveis

Podemos criar variáveis em python simplesmente atribuindo um valor a um nome possível de variável ainda não utilizado, da seguinte forma:

```
variavel = valor
```

Dessa forma, não precisamos utilizar nenhuma palavra chave para declarações de variáveis, como o "let" ou "const" do javascript

Os primeiros valores que vamos aprender são os numéricos e os textuais 

#### Criando uma variável numérica(int):
```
variavel = 10
```

#### Criando uma variável textual(str):
```
variavel = "Oi"
```

Agora que já sabemos como criar alguns tipo básicos, podemos aprender os diversos tipo de variáveis que podemos utilizar:

Os tipos de variáveis existente no python são:

- Tipo Vazio: NoneType
    - Tipo vazio quando algo não possui um valor definido
    - Pode ser criado utilizando variavel = None 
- Texto: str
    - Tipo textual
    - Pode ser criado com:
        - variavel = "texto"
- Números: int, float, complex
    - int:
        - valor númerico para números inteiros(sem virgula)
        - pode ser criado com:
            - variavel = 10 #um número qualquer
    - float:
        - variável que representa números de ponto flutuante
        - pode ser criado com:
            - numero = 2.2 #um número com ponto   OU
            - numero = 2e2 #número escrito em notação científica
    - complex:
        - número que contém uma parte imaginária
        - pode ser criádo com:
            - 2+2j  #parte real + parte imaginária
- Tipo Verdade: bool
    - Tipo usado para guardar verdadeiro e falso
    - Pode ser criado com:
        - verdade = True #Primeira letra TEM que ser maiúscula, senão não funciona
- Listas ordenadas: list, tuple, range, dict
    - Lista(list):  Coleção ordenada, mutável que permite valores duplicados
    - Tupla(tuple): Coleção ordenada, imutável e permite duplicatas
    - Range(range): Coleção sequencial de números; imutável
    - Dicionário(dict): Coleção ordenada, mutável composta de chave/valor; **permite** duplicatas de VALOR, **não permite** duplicatas de CHAVE
- Listas não ordenadas: set, frozenset
    - set: Conjunto não ordenado, parcialmente não mutável(permite adição e remoção de valores, não permite alteração de valores já no conjunto) que não permite duplicatas
    - frozenset: Conjunto não ordenado, não mutável, que não permite duplicatas

Outros tipos:
